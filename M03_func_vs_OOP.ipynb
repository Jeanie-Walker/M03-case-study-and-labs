```python
#Sort an Array of 0s, 1s, and 2s
class Solution:
    def sort012(self, arr):
        count_0 = 0
        count_1 = 0
        count_2 = 0
    
        for num in arr:
            if num == 0:
                count_0 += 1
                
            elif num == 1:
                count_1 += 1
                
            else:
                count_2 += 1
        
        
        position = 0
    
        for _ in range(count_0):
            arr[position] = 0
            position += 1
            
        for _ in range(count_1):
            arr[position] = 1
            position += 1
            
        for _ in range(count_2):
            arr[position] = 2
            position += 1
        
```

```python
#Binary Search
class Solution:
    def firstSearch(self, arr, k):
        low = 0
        high = len(arr) - 1
        answer = -1

        while low <= high:
            mid = low + (high - low) // 2

            if arr[mid] == k:
                answer = mid
                high = mid - 1
            elif arr[mid] < k:
                low = mid + 1
            else:
                high = mid - 1

        return answer
```
